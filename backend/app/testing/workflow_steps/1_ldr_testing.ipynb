```python

import scanpy as sc
import os

file_id = "clean_testing_file_2.h5ad"
UPLOAD_DIR = "../../../../persistent01"

file_path = os.path.join(UPLOAD_DIR, file_id)

adata = sc.read_h5ad(file_path)

adata
```

```python
for i in adata.uns_keys():
  del adata.uns[i]
  
for i in adata.obsm_keys():
  del adata.obsm[i]

for i in adata.varm_keys():
  del adata.varm[i]

for i in list(adata.obsp.keys()):
  del adata.obsp[i]

adata
```

```python
key = "test_1606"
```

```python
n_pcs = 30
n_neighbors = 15
```

```python
sc.pp.pca(
  adata,
  key_added=f"X_pca_{key}",
  n_comps=n_pcs,
  # chunked=True,
  # chunk_size=1000,
  # zero_center=True,
  # svd_solver='arpack',
)

adata
```

```python
sc.pp.neighbors(
  adata,
  key_added=f"neighbors_{key}",
  use_rep=f"X_pca_{key}",
  n_pcs=n_pcs,
  n_neighbors=n_neighbors
)
```
