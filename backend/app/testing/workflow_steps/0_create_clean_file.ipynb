```python
import scanpy as sc
import os

file_id = "clean_testing_file_2.h5ad"
UPLOAD_DIR = "../../../../persistent01"

file_path = os.path.join(UPLOAD_DIR, file_id)

adata = sc.read_h5ad(file_path)

adata
```

```python
for i in adata.uns_keys():
  del adata.uns[i]
  
for i in adata.obsm_keys():
  del adata.obsm[i]

for i in adata.varm_keys():
  del adata.varm[i]

for i in list(adata.obsp.keys()):
  del adata.obsp[i]

adata
```

```python
new_file_id = "clean_testing_file_2.h5ad"
new_file_path = os.path.join(UPLOAD_DIR, new_file_id)

sc.write(new_file_path, adata)
```

```python
adata.file.close()
```
